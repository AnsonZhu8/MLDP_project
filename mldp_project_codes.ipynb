```python
print("RestingBP = 0:", (df["RestingBP"] == 0).sum())
print("Cholesterol = 0:", (df["Cholesterol"] == 0).sum())
```

##### find out how many patients have restingbp, cholesterol values of 0 
##### indeed there is a patient that have restingbp as 0 which not possible
##### wow 225 patients with 0 cholesterol 
##### *FIX THOSE WITH MEDIAN* fixed with median to not get influenced by outliers

## fixing the features with median

```python
#replacing the restingbp and cholesterol that has value of 0 with the median if their respective variables
df["RestingBP"] = df["RestingBP"].replace(0, df["RestingBP"].median())
df["Cholesterol"] = df["Cholesterol"].replace(0, df["Cholesterol"].median())
```

```python
df.describe(include='all')
```

```python
print("RestingBP = 0:", (df["RestingBP"] == 0).sum())
print("Cholesterol = 0:", (df["Cholesterol"] == 0).sum())

#double check
```
